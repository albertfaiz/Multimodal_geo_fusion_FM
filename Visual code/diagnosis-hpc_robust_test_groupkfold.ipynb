Looking for Mann-Kendall

```python
import pandas as pd
import numpy as np
from scipy.stats import kendalltau
import warnings

warnings.filterwarnings("ignore")

# ====================================================================
# 1. LOAD DATA
# ====================================================================
CSV_FILE = ('/Users/faizahmad/Desktop/paper1/paper1_run1/full_clean_engineered_dataset_with_LE.csv')
print(f"📦 Loading dataset: {CSV_FILE}...")
df = pd.read_csv(CSV_FILE)

# Ensure 'year' column exists
year_col = 'year' if 'year' in df.columns else 'Year'

# ====================================================================
# 2. AGGREGATE BY YEAR (National Mean)
# ====================================================================
# We group by year and take the mean to test for systemic global drift
print("🔄 Aggregating national averages by year (2000-2019)...")
yearly_df = df.groupby(year_col).mean(numeric_only=True).reset_index()
years = yearly_df[year_col].values

# ====================================================================
# 3. DEFINE "USEFUL MATRICES" (Features to test)
# ====================================================================
# You can test all columns, but let's highlight the crucial ones from your paper
features_to_test = [
    # Top SHAP and core physics features
    "Nighttime Cooling Efficiency",
    "Nighttime Temp (10th Percentile)",
    "Thermal Vegetation Index (TVI)",
    "Daytime Surface Temp (Mean)",
    "Landsat NDWI (Mean)",
    
    # Let's dynamically find your NDVI and LST columns if the exact names above differ
]

# Add any column containing 'NDVI', 'LST', or 'Temp' to our test list automatically
for col in df.columns:
    if any(keyword in col.lower() for keyword in ['ndvi', 'lst', 'temp']):
        if col not in features_to_test:
            features_to_test.append(col)

# Keep only features that actually exist in the dataframe
valid_features = [f for f in features_to_test if f in yearly_df.columns]

# ====================================================================
# 4. RUN MANN-KENDALL (KENDALL'S TAU) TREND TEST
# ====================================================================
print(f"\n📈 MANN-KENDALL TREND TEST RESULTS (n={len(years)} years)")
print("=" * 80)
print(f"{'Feature Name':<45} | {'Tau (τ)':<8} | {'P-Value':<8} | {'Trend Status'}")
print("-" * 80)

results = []

for feat in valid_features:
    # Get the 20-year time series for this specific feature
    values = yearly_df[feat].values
    
    # Calculate Kendall's Tau and p-value
    tau, p_val = kendalltau(years, values)
    
    # Determine significance (p < 0.05 is standard for identifying a real trend)
    if p_val < 0.001:
        sig = "*** (Strong Trend)"
    elif p_val < 0.01:
        sig = "** (Moderate Trend)"
    elif p_val < 0.05:
        sig = "* (Weak Trend)"
    else:
        sig = "ns (No Significant Drift)"
        
    results.append({
        "Feature": feat,
        "Tau": tau,
        "P-value": p_val,
        "Status": sig
    })

# Sort results by absolute Tau (strongest trends at the top)
results_df = pd.DataFrame(results)
results_df['Abs_Tau'] = results_df['Tau'].abs()
results_df = results_df.sort_values(by='Abs_Tau', ascending=False).drop(columns=['Abs_Tau'])

# Print cleanly
for _, row in results_df.iterrows():
    print(f"{row['Feature']:<45} | {row['Tau']:>8.3f} | {row['P-value']:>8.4f} | {row['Status']}")

print("=" * 80)
print("💡 Interpretation Guide:")
print(" - Tau (τ) close to 0 means the satellite data is completely stable over 20 years.")
print(" - A high p-value (e.g., > 0.05) proves there is NO significant sensor degradation/drift.")
```

```python
import pickle
import pandas as pd
import numpy as np
from pathlib import Path
import warnings
from scipy.stats import linregress, skew
from statsmodels.nonparametric.smoothers_lowess import lowess
import re

warnings.filterwarnings("ignore")

# ====================================================================
# 0. PATHS & DATA LOADING
# ====================================================================
RUN_DIR = Path("/Users/faizahmad/Desktop/paper1 2026-04-18 /paper1_run1/")
P1_DIR  = RUN_DIR / "results_individual_modalities_final"
P2_DIR  = RUN_DIR / "results_production_final"

PKL_FILE      = P2_DIR / "ml_results.pkl"
CSV_FILE      = RUN_DIR / "full_clean_engineered_dataset_with_LE.csv"
MASTER_SUM    = P1_DIR / "master_summary_formatted.csv"
PREDICTIONS   = P2_DIR / "predictions_all_folds.csv"
COUNTY_SHP    = Path("/Users/faizahmad/Desktop/grok_live/data/cb_2018_us_county_500k/cb_2018_us_county_500k.shp")

print("🚀 LOADING GOD SCRIPT: DETAILED VERIFICATION & CROSS-CHECKING...")
print("="*80)

with open(PKL_FILE, "rb") as f:
    res = pickle.load(f)

df = pd.read_csv(CSV_FILE)
preds = pd.read_csv(PREDICTIONS) if PREDICTIONS.exists() else res["predictions"]
abl_df = pd.read_csv(MASTER_SUM) if MASTER_SUM.exists() else None

shap_imp = res["shap_importance"]

# --- FIX: Guarantee 'Rank' column exists ---
if "Rank" not in shap_imp.columns:
    shap_imp = shap_imp.sort_values(by="Mean_|SHAP|", ascending=False).reset_index(drop=True)
    shap_imp["Rank"] = shap_imp.index + 1
# -------------------------------------------

shap_vals = res["shap_values"]
X_sample = res["shap_sample"]
fn_list = list(X_sample.columns)
tables = res["tables"]

# Dictionary to store all LaTeX variables dynamically
latex_vars = {}

# ====================================================================
# INSPECTION 0: PRINT ALL AVAILABLE KEYS IN PICKLE
# ====================================================================
print("\n📦 PICKLE FILE CONTENTS (All Top-Level Keys):")
print("-" * 80)
for key in sorted(res.keys()):
    val_type = type(res[key]).__name__
    if isinstance(res[key], dict):
        print(f"  ✓ {key:30s} (dict with {len(res[key])} items)")
        for subkey in list(res[key].keys())[:5]:
            print(f"      └─ {subkey}")
        if len(res[key]) > 5:
            print(f"      └─ ... and {len(res[key]) - 5} more")
    elif isinstance(res[key], pd.DataFrame):
        print(f"  ✓ {key:30s} (DataFrame: {res[key].shape[0]} rows × {res[key].shape[1]} cols)")
    elif isinstance(res[key], np.ndarray):
        print(f"  ✓ {key:30s} (ndarray: {res[key].shape})")
    else:
        print(f"  ✓ {key:30s} ({val_type})")

# ====================================================================
# INSPECTION 1: COLUMN NAMES FROM X_SAMPLE (SHAP Features)
# ====================================================================
print("\n\n📊 FEATURE LIST FROM X_SAMPLE (Total: {})".format(len(fn_list)))
print("-" * 80)
for i, feat in enumerate(fn_list, 1):
    print(f"  {i:3d}. {feat}")

# ====================================================================
# INSPECTION 2: SHAP MATRIX DETAILS
# ====================================================================
print(f"\n\n🧮 SHAP VALUES MATRIX DETAILS:")
print("-" * 80)
print(f"  Shape: {shap_vals.shape}")
print(f"  Data type: {shap_vals.dtype}")
# make sure shap_vals supports reductions
try:
    print(f"  Min value: {shap_vals.min():.6f}")
    print(f"  Max value: {shap_vals.max():.6f}")
    print(f"  Mean value: {shap_vals.mean():.6f}")
    print(f"  Std Dev: {shap_vals.std():.6f}")
    print(f"  Total NaNs: {np.isnan(shap_vals).sum()}")
    print(f"  Total Infs: {np.isinf(shap_vals).sum()}")
except Exception as e:
    print("  ⚠️  Could not compute some SHAP statistics:", e)

# ====================================================================
# INSPECTION 3: SHAP IMPORTANCE DETAILS
# ====================================================================
print(f"\n\n📈 SHAP IMPORTANCE DATAFRAME:")
print("-" * 80)
print(f"  Shape: {shap_imp.shape}")
print(f"  Columns: {list(shap_imp.columns)}")
print(f"\n  Top 10 features by Mean_|SHAP|:")
top10 = shap_imp.nlargest(10, "Mean_|SHAP|")
for i, (idx, row) in enumerate(top10.iterrows(), 1):
    print(f"    {i:2d}. {row['Feature']:45s} | Mean|SHAP|: {row['Mean_|SHAP|']:.6f}")

# ====================================================================
# INSPECTION 4: MODEL HYPERPARAMETERS (if available)
# ====================================================================
print(f"\n\n⚙️  MODEL HYPERPARAMETERS & CONFIGURATION:")
print("-" * 80)
if "model_params" in res:
    print("  Model Parameters found:")
    for k, v in res["model_params"].items():
        print(f"    {k}: {v}")
elif "rf_params" in res:
    print("  RF Parameters found:")
    for k, v in res["rf_params"].items():
        print(f"    {k}: {v}")
else:
    print("  ⚠️  No explicit hyperparameters stored. Checking tables for clues...")
    if "table1_cv_performance" in tables:
        print("  CV Performance table found (see Table 1 details below)")

# ====================================================================
# SECTION 1: OVERALL PERFORMANCE & FOLDS
# ====================================================================
print(f"\n\n{'='*80}")
print("SECTION 1: OVERALL PERFORMANCE & FOLDS")
print("="*80)

cv = tables["table1_cv_performance"]
print("\nCross-Validation Performance Table:")
print(cv.to_string())

def get_cv_stat(metric):
    val_str = cv.loc[cv["Fold"] == "Mean ± SD", metric].values[0]
    parts = str(val_str).split(" ± ")
    return float(parts[0]), float(parts[1])

r2, r2sd = get_cv_stat("Test_R2")
mae, maesd = get_cv_stat("MAE")
rmse, rmsesd = get_cv_stat("RMSE")

print(f"\n✅ Mean Metrics Extracted:")
print(f"  R² = {r2:.4f} ± {r2sd:.4f}")
print(f"  MAE = {mae:.4f} ± {maesd:.4f}")
print(f"  RMSE = {rmse:.4f} ± {rmsesd:.4f}")

latex_vars.update({"CombRsq": f"{r2:.3f}", "CombRsqSD": f"{r2sd:.3f}", 
                   "CombMAE": f"{mae:.2f}", "CombMAESD": f"{maesd:.2f}", 
                   "CombRMSE": f"{rmse:.2f}", "CombRMSESD": f"{rmsesd:.2f}"})

# Individual Folds
print(f"\n✅ Individual Fold Metrics:")
for i in range(1, 6):
    row = cv.loc[cv["Fold"] == str(i)]
    if not row.empty:
        fold_r2 = float(row['Test_R2'].values[0])
        fold_mae = float(row['MAE'].values[0])
        fold_rmse = float(row['RMSE'].values[0])
        word = ["One", "Two", "Three", "Four", "Five"][i-1]
        print(f"  Fold {i}: R²={fold_r2:.4f}, MAE={fold_mae:.4f}, RMSE={fold_rmse:.4f}")
        latex_vars.update({
            f"Fold{word}Rsq": f"{fold_r2:.3f}",
            f"Fold{word}MAE": f"{fold_mae:.2f}",
            f"Fold{word}RMSE": f"{fold_rmse:.2f}"
        })

# ====================================================================
# SECTION 2: ABLATION & FUSION GAIN
# ====================================================================
print(f"\n\n{'='*80}")
print("SECTION 2: ABLATION & FUSION GAIN")
print("="*80)

if abl_df is not None:
    print("\nAblation Study Results:")
    print(abl_df.to_string())
    
    for idx, row in abl_df.iterrows():
        mod = row["Modality"]
        r2_str = str(row["R2"]).split(" ± ")
        mae_str = str(row["MAE (years)"]).split(" ± ")
        
        prefix = "AblComb" if mod == "Combined" else "AblEng" if mod == "Engineered" else \
                 "AblLST" if mod == "LST" else "AblUSDA" if mod == "USDA" else \
                 "AblLive" if mod == "Livestock" else "AblNDVI" if mod == "NDVI" else \
                 "AblDEM" if mod == "DEM" else "AblLandsat" if mod == "Landsat" else \
                 "AblSAR" if mod == "S1" else "AblSoil" if mod == "Soil" else \
                 "AblSTwo" if mod == "S2" else "AblJRC" if mod == "JRC" else ""
        
        if prefix:
            print(f"  {mod:15s}: R²={float(r2_str[0]):.4f}, MAE={float(mae_str[0]):.4f} yrs")
            latex_vars[f"{prefix}Rsq"] = f"{float(r2_str[0]):.3f}"
            latex_vars[f"{prefix}MAE"] = f"{float(mae_str[0]):.2f}"

    # Fusion Gain Calculation
    comb_r2 = float(latex_vars.get("AblCombRsq", r2))
    lst_r2 = float(latex_vars.get("AblLSTRsq", 0.429))
    gain = (comb_r2 - lst_r2) / (1 - lst_r2) * 100
    
    print(f"\n✅ Fusion Gain Calculation:")
    print(f"  Combined R² = {comb_r2:.4f}")
    print(f"  LST R² = {lst_r2:.4f}")
    print(f"  Fusion Gain = ({comb_r2:.4f} - {lst_r2:.4f}) / (1 - {lst_r2:.4f}) × 100")
    print(f"  Fusion Gain = {gain:.2f}%")
    latex_vars["FusionGain"] = f"{gain:.0f}"
else:
    print("⚠️  Ablation data not available")

# ====================================================================
# SECTION 3: NIGHTTIME PARADOX (DETAILED)
# ====================================================================
print(f"\n\n{'='*80}")
print("SECTION 3: NIGHTTIME PARADOX - DETAILED BREAKDOWN")
print("="*80)

print("\n📋 Searching for Nighttime-related features...")
night_mask = shap_imp["Feature"].str.contains(r"Nighttime|Cooling Efficiency|Night Temp", case=False, na=False)
night_features = shap_imp.loc[night_mask]

print(f"\n  Found {len(night_features)} Nighttime Features:")
for i, (idx, row) in enumerate(night_features.iterrows(), 1):
    print(f"    {i}. {row['Feature']:50s} | Mean|SHAP|: {row['Mean_|SHAP|']:10.6f} | Rank: {row['Rank']}")

night_sum = night_features["Mean_|SHAP|"].sum()
print(f"\n  Total Nighttime SHAP Importance: {night_sum:.6f}")

print("\n📋 Searching for Daytime-related features...")
day_mask = shap_imp["Feature"].str.contains(r"Daytime|Day Temp", case=False, na=False)
day_features = shap_imp.loc[day_mask]

print(f"\n  Found {len(day_features)} Daytime Features:")
for i, (idx, row) in enumerate(day_features.iterrows(), 1):
    print(f"    {i}. {row['Feature']:50s} | Mean|SHAP|: {row['Mean_|SHAP|']:10.6f} | Rank: {row['Rank']}")

day_sum = day_features["Mean_|SHAP|"].sum()
print(f"\n  Total Daytime SHAP Importance: {day_sum:.6f}")

nce_mask = shap_imp["Feature"].str.contains("Nighttime Cooling Efficiency", na=False)
nce_rows = shap_imp.loc[nce_mask]
if not nce_rows.empty:
    nce_val = nce_rows["Mean_|SHAP|"].values[0]
    nce_rank = nce_rows["Rank"].values[0]
    print(f"\n🌙 Nighttime Cooling Efficiency (NCE):")
    print(f"  Mean|SHAP|: {nce_val:.6f}")
    print(f"  Rank: {nce_rank}")
else:
    nce_val = np.nan
    print("  ⚠️  NCE not found in features")

night_day_ratio = night_sum / day_sum if day_sum > 0 else np.nan
print(f"\n✅ Night/Day SHAP Ratio: {night_sum:.6f} / {day_sum:.6f} = {night_day_ratio:.2f}")

latex_vars.update({
    "SHAPnightTotal": f"{night_sum:.3f}", 
    "SHAPdayTotal": f"{day_sum:.3f}",
    "NightDayRatio": f"{night_day_ratio:.1f}", 
    "SHAPnce": f"{nce_val:.3f}"
})

# ====================================================================
# SECTION 4: TEMPORAL STABILITY
# ====================================================================
print(f"\n\n{'='*80}")
print("SECTION 4: TEMPORAL STABILITY (Year 2000 & 20-Year Trends)")
print("="*80)

t3 = tables.get("table3_temporal", None)
if t3 is not None:
    print("\nTable 3 - Temporal Performance:")
    print(t3.to_string())
    
    # Year 2000 Analysis
    if isinstance(t3.index, pd.Index) and 2000 in t3.index:
        y2k = t3.loc[2000]
        print(f"\n✅ Year 2000 Baseline:")
        print(f"  R² = {y2k['R2']:.4f}")
        print(f"  MAE = {y2k['MAE']:.4f}")
        print(f"  RMSE = {y2k['RMSE']:.4f}")
        print(f"  Counties = {int(y2k['Counties']):,}")
        latex_vars.update({
            "YearZeroRsq": f"{y2k['R2']:.3f}", 
            "YearZeroMAE": f"{y2k['MAE']:.3f}", 
            "YearZeroN": f"{int(y2k['Counties']):,}"
        })
    
    # OLS Trends
    years = t3.index.values
    r2_vals = t3["R2"].values
    mae_vals = t3["MAE"].values
    rmse_vals = t3["RMSE"].values
    
    sl_r2, int_r2, r_r2, p_r2, se_r2 = linregress(years, r2_vals)
    sl_mae, int_mae, r_mae, p_mae, se_mae = linregress(years, mae_vals)
    sl_rmse, int_rmse, r_rmse, p_rmse, se_rmse = linregress(years, rmse_vals)
    
    print(f"\n✅ 20-Year Trend Analysis (OLS Linear Regression):")
    print(f"\n  R² Trend:")
    print(f"    Slope (β): {sl_r2:+.6f} per year")
    print(f"    Intercept: {int_r2:.4f}")
    print(f"    Correlation (r): {r_r2:.4f}")
    print(f"    P-value: {p_r2:.6f} {'***' if p_r2 < 0.001 else '**' if p_r2 < 0.01 else '*' if p_r2 < 0.05 else ''}")
    print(f"    Std Error: {se_r2:.6f}")
    print(f"    Interpretation: R² {'declining' if sl_r2 < 0 else 'improving'} by {abs(sl_r2):.6f} per year")
    
    print(f"\n  MAE Trend:")
    print(f"    Slope (β): {sl_mae:+.6f} per year")
    print(f"    Intercept: {int_mae:.4f}")
    print(f"    Correlation (r): {r_mae:.4f}")
    print(f"    P-value: {p_mae:.6f} {'***' if p_mae < 0.001 else '**' if p_mae < 0.01 else '*' if p_mae < 0.05 else ''}")
    print(f"    Std Error: {se_mae:.6f}")
    
    print(f"\n  RMSE Trend:")
    print(f"    Slope (β): {sl_rmse:+.6f} per year")
    print(f"    Intercept: {int_rmse:.4f}")
    print(f"    Correlation (r): {r_rmse:.4f}")
    print(f"    P-value: {p_rmse:.6f} {'***' if p_rmse < 0.001 else '**' if p_rmse < 0.01 else '*' if p_rmse < 0.05 else ''}")
    print(f"    Std Error: {se_rmse:.6f}")
    
    latex_vars.update({
        "TemporalRsqBeta": f"{sl_r2:+.4f}", 
        "TemporalRsqP": f"{p_r2:.2f}",
        "TemporalMAEBeta": f"{sl_mae:+.4f}", 
        "TemporalRMSEBeta": f"{sl_rmse:+.4f}"
    })
else:
    print("⚠️  Table 3 (temporal) not found")

# ====================================================================
# SECTION 5: BIAS & RESIDUALS
# ====================================================================
print(f"\n\n{'='*80}")
print("SECTION 5: BIAS & RESIDUALS ANALYSIS")
print("="*80)

preds["fips"] = preds["fips"].astype(str).str.zfill(5)
res_by_fips = preds.groupby("fips")["residual"].mean()

# Worst Over-predicted & Under-predicted
over = res_by_fips.nsmallest(5)
under = res_by_fips.nlargest(2)

print(f"\n✅ Residual Statistics by County:")
print(f"  Worst Over-predicted (Top 5 - most negative):")
for fips, val in over.items():
    print(f"    FIPS {fips}: {val:+.4f} years")

print(f"\n  Worst Under-predicted (Top 2 - most positive):")
for fips, val in under.items():
    print(f"    FIPS {fips}: {val:+.4f} years")

latex_vars.update({
    "BiasOglala": f"{over.min():.2f}", 
    "BiasHighA": f"{under.max():+.2f}"
})

# Overall Residual Distribution
res_arr = preds["residual"].values
res_mean = res_arr.mean()
res_std = res_arr.std()
res_skew = skew(res_arr)
res_min = res_arr.min()
res_max = res_arr.max()
res_q25 = np.percentile(res_arr, 25)
res_q50 = np.percentile(res_arr, 50)
res_q75 = np.percentile(res_arr, 75)

print(f"\n✅ Overall Residual Distribution (n={len(res_arr):,}):")
print(f"  Mean: {res_mean:+.6f}")
print(f"  Std Dev: {res_std:.6f}")
print(f"  Skewness: {res_skew:.4f} {'(right-skewed)' if res_skew > 0 else '(left-skewed)'}")
print(f"  Min: {res_min:.4f}")
print(f"  Q25: {res_q25:.4f}")
print(f"  Median (Q50): {res_q50:.4f}")
print(f"  Q75: {res_q75:.4f}")
print(f"  Max: {res_max:.4f}")
print(f"  IQR: {res_q75 - res_q25:.4f}")

latex_vars.update({
    "ResidMean": f"{res_mean:.3f}", 
    "ResidSkew": f"{res_skew:.2f}",
    "ResidMin": f"{res_min:.2f}", 
    "ResidMax": f"{res_max:.2f}"
})

# ====================================================================
# SECTION 6: MORAN'S I (Spatial Autocorrelation)
# ====================================================================
print(f"\n\n{'='*80}")
print("SECTION 6: SPATIAL AUTOCORRELATION (Moran's I)")
print("="*80)

try:
    import geopandas as gpd
    import libpysal
    from esda import moran
    
    print("\n🗺️  Loading county shapefile...")
    gdf = gpd.read_file(COUNTY_SHP)
    
    # Filter for contiguous US states (exclude Alaska, Hawaii, and territories)
    contiguous_states = [f'{i:02d}' for i in range(1, 57) if i not in [2, 15, 60, 66, 69, 72, 78]]
    gdf = gdf[gdf['STATEFP'].isin(contiguous_states)]
    
    gdf["fips"] = gdf["GEOID" if "GEOID" in gdf.columns else "geoid"].astype(str).str.zfill(5)
    
    print(f"  Shapefile counties (contiguous US): {len(gdf):,}")
    
    # Merge with residuals from ML output
    gdf = gdf.merge(res_by_fips.rename("res_mean"), left_on="fips", right_index=True, how="inner")
    
    print(f"  Counties with residual data: {len(gdf):,}")
    print(f"  Unique counties in analysis: {gdf['fips'].nunique()}")
    
    if len(gdf) < 10:
        print(f"  ⚠️  Too few counties ({len(gdf)}) for spatial analysis. Skipping Moran's I.")
    else:
        # Create Queen contiguity weights (only for counties that have data)
        w = libpysal.weights.Queen.from_dataframe(gdf)
        
        # Check for islands (counties with no neighbors)
        islands = [node for node in w.neighbors if len(w.neighbors[node]) == 0]
        if len(islands) > 0:
            print(f"\n  ⚠️  Found {len(islands)} island counties (no neighbors)")
            gdf_valid = gdf[~gdf.index.isin(islands)].copy()
            w_valid = libpysal.weights.Queen.from_dataframe(gdf_valid)
        else:
            gdf_valid = gdf.copy()
            w_valid = w
        
        print(f"  Counties used for Moran's I: {len(gdf_valid):,}")
        
        # Row-standardize weights
        w_valid.transform = 'r'
        
        # Calculate Moran's I with permutation test
        mi = moran.Moran(gdf_valid["res_mean"].values, w_valid, permutations=999)
        
        # --- BULLETPROOF ATTRIBUTE EXTRACTION ---
        # Safely checks multiple common esda version names, defaults to "N/A" or 0 if missing
        variance = getattr(mi, 'VI_norm', getattr(mi, 'v_norm', getattr(mi, 'VI', 'N/A')))
        z_score  = getattr(mi, 'z_norm', getattr(mi, 'z', 0.0))
        p_norm   = getattr(mi, 'p_norm', getattr(mi, 'p', 1.0))
        p_sim    = getattr(mi, 'p_sim', 1.0)
        perms    = getattr(mi, 'permutations', getattr(mi, 'n_permutations', 999))
        
        print(f"\n✅ Moran's I Spatial Autocorrelation Test:")
        print(f"  Moran's I: {mi.I:.6f}")
        print(f"  E[I] (expected under null): {mi.EI:.6f}")
        
        if isinstance(variance, float):
            print(f"  Variance (normal approx): {variance:.6f}")
        else:
            print(f"  Variance: {variance}")
            
        print(f"  Z-score (normal approx): {z_score:.4f}")
        print(f"  P-value (normal approx): {p_norm:.6f}")
        print(f"  P-value (permutation-based): {p_sim:.6f}")
        print(f"  Permutations performed: {perms}")
        
        significance = "***" if p_sim < 0.001 else "**" if p_sim < 0.01 else "*" if p_sim < 0.05 else "ns"
        
        if p_sim < 0.05:
            direction = "positive" if mi.I > mi.EI else "negative"
            print(f"  ✓ SIGNIFICANT {direction} spatial autocorrelation (p={p_sim:.4f} {significance})")
        else:
            print(f"  ◯ No significant spatial autocorrelation (p={p_sim:.4f} {significance})")
        
        latex_vars.update({
            "MoranI": f"{mi.I:.4f}", 
            "MoranP": f"{p_sim:.4f}",
            "MoranZ": f"{z_score:.4f}",
            "MoranNperms": f"{perms}"
        })
    
except ImportError as e:
    print(f"⚠️  Missing required package: {e}")
except Exception as e:
    print(f"⚠️  Moran's I calculation failed: {type(e).__name__}: {e}")

# ====================================================================
# SECTION 7: LOWESS INFLECTION POINTS (Top 10 Features)
# ====================================================================
print(f"\n\n{'='*80}")
print("SECTION 7: LOWESS CURVE ANALYSIS - INFLECTION POINTS (Top 10 Features)")
print("="*80)

def get_lowess_inflection_detailed(feat_name, find_max=False, skip_start_frac=0.05):
    """Calculate LOWESS trend and find inflection point with detailed output.
       - skip_start_frac: fraction of sorted x to ignore at the low end to avoid noisy startup.
    """
    if feat_name not in fn_list:
        return np.nan, None, None, None
    
    idx = fn_list.index(feat_name)
    x = np.asarray(X_sample[feat_name].values, dtype=float)
    y = np.asarray(shap_vals[:, idx], dtype=float)
    
    # Remove NaNs and Infs
    valid_mask = np.isfinite(x) & np.isfinite(y)
    x_clean = x[valid_mask]
    y_clean = y[valid_mask]
    
    if len(x_clean) < 10:
        return np.nan, None, None, None
    
    # Sort and apply LOWESS
    order = np.argsort(x_clean)
    lw = lowess(y_clean[order], x_clean[order], frac=0.20, return_sorted=True)
    lx, ly = lw[:, 0], lw[:, 1]
    
    if find_max:
        peak_idx = np.argmax(ly)
        return float(lx[peak_idx]), lx, ly, "max"
    
    # skip early noisy portion
    skip_n = int(len(lx) * skip_start_frac)
    if skip_n >= len(lx) - 2:
        skip_n = max(0, len(lx) - 3)
    idx_range = np.arange(skip_n, len(lx))
    
    # derivative and second derivative
    dy = np.gradient(ly, lx)
    ddy = np.gradient(dy, lx)
    
    # Inflection via change in sign of second derivative (after skip)
    sign_changes = idx_range[:-1][np.where(np.sign(ddy[idx_range[:-1]]) != np.sign(ddy[idx_range[1:]]))[0]]
    if len(sign_changes) > 0:
        inflection_idx = sign_changes[0]
        return float(lx[inflection_idx]), lx, ly, "inflection"
    
    # Fallback: find where first derivative crosses zero (after skip)
    zero_cross = idx_range[np.where(np.diff(np.sign(dy[idx_range])) != 0)[0]] if len(dy[idx_range])>1 else []
    if len(zero_cross) > 0:
        return float(lx[zero_cross[0]]), lx, ly, "deriv_zero"
    
    return np.nan, lx, ly, None

print("\n📊 Top 10 SHAP Features - Inflection Point Analysis:")
top10_features = shap_imp.nlargest(10, "Mean_|SHAP|")

for i, (idx, row) in enumerate(top10_features.iterrows(), 1):
    feat_name = row["Feature"]
    feat_rank = row["Rank"]
    feat_shap = row["Mean_|SHAP|"]
    
    inflection_val, lx, ly, inflection_type = get_lowess_inflection_detailed(feat_name, find_max=False)
    
    print(f"\n  {i}. {feat_name} (Rank: {feat_rank}, Mean|SHAP|: {feat_shap:.6f})")
    
    if not np.isnan(inflection_val):
        print(f"     Inflection Point: {inflection_val:.4f}")
        if lx is not None and ly is not None:
            min_x, max_x = lx.min(), lx.max()
            min_y, max_y = ly.min(), ly.max()
            print(f"     X range: [{min_x:.4f}, {max_x:.4f}]")
            print(f"     Y range: [{min_y:.6f}, {max_y:.6f}]")
    else:
        print(f"     ⚠️  Could not determine inflection point")

# ====================================================================
# SECTION 8: PHYSICAL THRESHOLDS
# ====================================================================
print(f"\n\n{'='*80}")
print("SECTION 8: PHYSICAL THRESHOLDS (Forest, Soil, LST, Cattle)")
print("="*80)

# Forest Buffering (LST & Forest interaction)
LST_FEAT = "Daytime Surface Temp (Mean)"
FOREST_FEAT = "Deciduous Forest %"

print(f"\n🌲 Forest Buffering Analysis:")
print(f"  LST Feature: {LST_FEAT}")
print(f"  Forest Feature: {FOREST_FEAT}")

if LST_FEAT in fn_list and FOREST_FEAT in fn_list:
    lst_idx = fn_list.index(LST_FEAT)
    
    lst_x = np.asarray(X_sample[LST_FEAT].values)
    lst_shap = np.asarray(shap_vals[:, lst_idx])
    forest_x = np.asarray(X_sample[FOREST_FEAT].values)
    
    print(f"  Found {len(X_sample)} samples for analysis")
    
    # Bin the data
    tq = pd.qcut(lst_x, q=4, labels=["Q1", "Q2", "Q3", "Q4"])
    fcat = pd.cut(forest_x, bins=[-np.inf, 0.05, 0.20, np.inf], 
                  labels=["Low (<5%)", "Medium (5-20%)", "High (>20%)"])
    
    df_i = pd.DataFrame({"tq": tq, "fc": fcat, "sv": lst_shap}).dropna()
    
    print(f"\n  Cross-tabulation of LST Quartiles × Forest Categories:")
    crosstab = pd.crosstab(df_i["tq"], df_i["fc"], margins=True)
    print(crosstab)
    
    # Calculate medians in the hot quartile
    q4l = df_i[(df_i["fc"] == "Low (<5%)") & (df_i["tq"] == "Q4")]["sv"].median()
    q4h = df_i[(df_i["fc"] == "High (>20%)") & (df_i["tq"] == "Q4")]["sv"].median()
    q4m = df_i[(df_i["fc"] == "Medium (5-20%)") & (df_i["tq"] == "Q4")]["sv"].median()
    
    print(f"\n  Q4 (Hottest Quartile) SHAP Values:")
    print(f"    Low Forest (<5%): {q4l:.6f}")
    print(f"    Medium Forest (5-20%): {q4m:.6f}")
    print(f"    High Forest (>20%): {q4h:.6f}")
    
    gap = q4l - q4h
    attenuation = abs(gap / q4l) * 100 if q4l != 0 else 0
    
    print(f"\n  ✅ Forest Buffering Effect:")
    print(f"    Gap (Low - High): {gap:+.6f}")
    print(f"    Attenuation: {attenuation:.2f}%")
    
    latex_vars.update({
        "ForestAtten": f"{attenuation:.0f}", 
        "ForestDelta": f"{gap:.3f}"
    })
else:
    print(f"  ⚠️  Missing required features")

# Soil Peak Analysis
soil_col = "Soil_mean" if "Soil_mean" in df.columns else \
           "Soil moisture mean" if "Soil moisture mean" in df.columns else None

print(f"\n🌾 Soil Moisture Optimal Band Analysis:")
print(f"  Soil Column: {soil_col}")

if soil_col and "MeanLifeExpectency" in df.columns:
    df_s = df[[soil_col, "MeanLifeExpectency"]].dropna()
    
    # Remove extreme outliers
    q_low = df_s[soil_col].quantile(0.01)
    q_high = df_s[soil_col].quantile(0.99)
    df_s = df_s[(df_s[soil_col] > q_low) & (df_s[soil_col] < q_high)]
    
    print(f"  Samples after cleaning: {len(df_s):,}")
    print(f"  Soil range: [{q_low:.2f}, {q_high:.2f}]")
    print(f"  LE range: [{df_s['MeanLifeExpectency'].min():.2f}, {df_s['MeanLifeExpectency'].max():.2f}]")
    
    # LOWESS fit
    order = np.argsort(df_s[soil_col].values)
    lw = lowess(df_s["MeanLifeExpectency"].values[order], df_s[soil_col].values[order], 
                frac=0.20, return_sorted=True)
    lx, ly = lw[:, 0], lw[:, 1]
    
    soil_peak_idx = np.argmax(ly)
    soil_peak = float(lx[soil_peak_idx])
    soil_peak_le = float(ly[soil_peak_idx])
    
    print(f"\n  ✅ Optimal Soil Moisture Band:")
    print(f"    Peak at: {soil_peak:.2f}")
    print(f"    Peak LE: {soil_peak_le:.2f} years")
    print(f"    LOWESS Min LE: {ly.min():.2f}, Max LE: {ly.max():.2f}")
    
    latex_vars["ThrSoilPeak"] = f"{soil_peak:.1f}"
else:
    print("  ⚠️  Could not analyze soil moisture")

# Temperature Thresholds from SHAP
print(f"\n🌡️  Temperature Inflection Points:")

night_feat = "Nighttime Temp (10th Percentile)"
day_feat = "Daytime Temp (10th Percentile)"

night_inflection, _, _, _ = get_lowess_inflection_detailed(night_feat, find_max=False)
day_inflection, _, _, _ = get_lowess_inflection_detailed(day_feat, find_max=False)

if not np.isnan(night_inflection):
    print(f"  {night_feat}: {night_inflection:.2f}°C")
    latex_vars["ThrNightLow"] = f"{night_inflection:.1f}"
else:
    print(f"  ⚠️  {night_feat} not found")

if not np.isnan(day_inflection):
    print(f"  {day_feat}: {day_inflection:.2f}°C")
    latex_vars["ThrDayLow"] = f"{day_inflection:.1f}"
else:
    print(f"  ⚠️  {day_feat} not found")


# ====================================================================
# FINAL OUTPUT: LATEX CONSTANTS
# ====================================================================
print(f"\n\n{'='*80}")
print("✅ FINAL LATEX CONSTANTS (COPY & PASTE INTO OVERLEAF)")
print("="*80)

for k, v in sorted(latex_vars.items()):
    print(f"\\newcommand{{\\{k}}}{{{v}}}")

print(f"\n{'='*80}")
print(f"Total Constants Generated: {len(latex_vars)}")
print("="*80)
```

```python
import pandas as pd
import numpy as np
from pathlib import Path
import sys
import warnings
from scipy.interpolate import interp1d
from scipy.stats import linregress, skew
from statsmodels.nonparametric.smoothers_lowess import lowess

warnings.filterwarnings("ignore")

# ====================================================================
# 0. PATHS & DATA LOADING (Combined from both code blocks)
# ====================================================================
# Paths adjusted for consistency
RESULTS_DIR = Path("/Users/faizahmad/Desktop/paper1/paper1_run1/results_production_final")
PKL_FILE = RESULTS_DIR / "ml_results.pkl"
CSV_FILE = Path("/Users/faizahmad/Desktop/paper1/paper1_run1/full_clean_engineered_dataset_with_LE.csv")
COUNTY_SHP = Path("/Users/faizahmad/Desktop/grok_live/data/cb_2018_us_county_500k/cb_2018_us_county_500k.shp")

print("🚀 LOADING COMBINED SCRIPT: DETAILED VERIFICATION & CROSS-CHECKING...")
print("="*80)

# Load pickle (from initial cell)
if not PKL_FILE.exists():
    print(f"❌ Error: Pickle file not found at {PKL_FILE}")
    sys.exit(1)

print(f"✅ Successfully located: {PKL_FILE.name}")
res = pd.read_pickle(PKL_FILE)

# Load CSV if available
df = pd.read_csv(CSV_FILE) if CSV_FILE.exists() else None
if df is None:
    print("⚠️ CSV file not found; some analyses will be skipped.")

# Extract key components from pickle
shap_imp = res.get("shap_importance")
shap_vals = res.get("shap_values")
X_sample = res.get("shap_sample")
fn_list = list(X_sample.columns) if X_sample is not None else []
tables = res.get("tables", {})
predictions = res.get("predictions")
cv_results = res.get("cv_results", [])

# Ensure shap_imp has Rank column (from later code)
if shap_imp is not None and "Rank" not in shap_imp.columns:
    shap_imp = shap_imp.sort_values(by="Mean_|SHAP|", ascending=False).reset_index(drop=True)
    shap_imp["Rank"] = shap_imp.index + 1

# ====================================================================
# 1. INSPECTION & METADATA (From initial cell, expanded)
# ====================================================================
print("\n📦 PICKLE FILE CONTENTS (All Top-Level Keys):")
print("-" * 80)
for key in sorted(res.keys()):
    val_type = type(res[key]).__name__
    if isinstance(res[key], dict):
        print(f"  ✓ {key:30s} (dict with {len(res[key])} items)")
        for subkey in list(res[key].keys())[:5]:
            print(f"      └─ {subkey}")
        if len(res[key]) > 5:
            print(f"      └─ ... and {len(res[key]) - 5} more")
    elif isinstance(res[key], pd.DataFrame):
        print(f"  ✓ {key:30s} (DataFrame: {res[key].shape[0]} rows × {res[key].shape[1]} cols)")
    elif isinstance(res[key], np.ndarray):
        print(f"  ✓ {key:30s} (ndarray: {res[key].shape})")
    else:
        print(f"  ✓ {key:30s} ({val_type})")

print(f"\n\n📊 FEATURE LIST FROM X_SAMPLE (Total: {len(fn_list)})")
print("-" * 80)
for i, feat in enumerate(fn_list, 1):
    print(f"  {i:3d}. {feat}")

print(f"\n\n🧮 SHAP VALUES MATRIX DETAILS:")
print("-" * 80)
if shap_vals is not None:
    print(f"  Shape: {shap_vals.shape}")
    print(f"  Data type: {shap_vals.dtype}")
    print(f"  Min value: {shap_vals.min():.6f}")
    print(f"  Max value: {shap_vals.max():.6f}")
    print(f"  Mean value: {shap_vals.mean():.6f}")
    print(f"  Std Dev: {shap_vals.std():.6f}")
    print(f"  Total NaNs: {np.isnan(shap_vals).sum()}")
    print(f"  Total Infs: {np.isinf(shap_vals).sum()}")

print(f"\n\n📈 SHAP IMPORTANCE DATAFRAME:")
print("-" * 80)
if shap_imp is not None:
    print(f"  Shape: {shap_imp.shape}")
    print(f"  Columns: {list(shap_imp.columns)}")
    print(f"\n  Top 10 features by Mean_|SHAP|:")
    top10 = shap_imp.nlargest(10, "Mean_|SHAP|")
    for i, (idx, row) in enumerate(top10.iterrows(), 1):
        print(f"    {i:2d}. {row['Feature']:45s} | Mean|SHAP|: {row['Mean_|SHAP|']:.6f}")

# ====================================================================
# 2. SPATIAL LIBRARIES & CONTIGUOUS US FILTER
# ====================================================================
try:
    import geopandas as gpd
    import libpysal
    import esda
    HAS_SPATIAL = True
    print("✅ Spatial libraries loaded.")
except ImportError:
    HAS_SPATIAL = False
    print("⚠️ Spatial libraries not found; Moran's I skipped.")

# Contiguous US state FIPS (exclude AK, HI, territories)
contiguous_states = [f'{i:02d}' for i in range(1, 57) if i not in [2, 15, 60, 66, 69, 72, 78]]

# ====================================================================
# 3. OVERALL PERFORMANCE & FOLDS (From later code, detailed)
# ====================================================================
print(f"\n\n{'='*80}")
print("SECTION 1: OVERALL PERFORMANCE & FOLDS")
print("="*80)

cv = tables.get("table1_cv_performance")
if cv is not None:
    print("\nCross-Validation Performance Table (Source: res['tables']['table1_cv_performance']):")
    print(cv.to_string())
    
    def get_cv_stat(metric):
        val_str = cv.loc[cv["Fold"] == "Mean ± SD", metric].values[0]
        parts = str(val_str).split(" ± ")
        return float(parts[0]), float(parts[1])
    
    r2, r2sd = get_cv_stat("Test_R2")
    mae, maesd = get_cv_stat("MAE")
    rmse, rmsesd = get_cv_stat("RMSE")
    
    print(f"\n✅ Mean Metrics Extracted (Parsed from string in table):")
    print(f"  R² = {r2:.4f} ± {r2sd:.4f}")
    print(f"  MAE = {mae:.4f} ± {maesd:.4f} years")
    print(f"  RMSE = {rmse:.4f} ± {rmsesd:.4f} years")
    
    # Individual Folds
    print(f"\n✅ Individual Fold Metrics (Direct from table):")
    for i in range(1, 6):
        row = cv.loc[cv["Fold"] == str(i)]
        if not row.empty:
            fold_r2 = float(row['Test_R2'].values[0])
            fold_mae = float(row['MAE'].values[0])
            fold_rmse = float(row['RMSE'].values[0])
            print(f"  Fold {i}: R²={fold_r2:.4f}, MAE={fold_mae:.4f}, RMSE={fold_rmse:.4f}")
else:
    print("⚠️ CV table not found.")

# Predictions-based metrics (from later code)
if predictions is not None:
    y_true = predictions["actual"].values
    y_pred = predictions["predicted"].values
    r2_pred = 1 - np.sum((y_true - y_pred) ** 2) / np.sum((y_true - y_true.mean()) ** 2)
    print(f"\nPredictions-based R² (full dataset, calculated from res['predictions']): {r2_pred:.4f}")

# ====================================================================
# 4. NIGHTTIME PARADOX (Detailed, from later code)
# ====================================================================
print(f"\n\n{'='*80}")
print("SECTION 2: NIGHTTIME PARADOX - DETAILED BREAKDOWN")
print("="*80)

if shap_imp is not None:
    night_mask = shap_imp["Feature"].str.contains(r"Nighttime|Cooling Efficiency|Night Temp", case=False, na=False)
    night_features = shap_imp.loc[night_mask]
    night_sum = night_features["Mean_|SHAP|"].sum()
    
    day_mask = shap_imp["Feature"].str.contains(r"Daytime|Day Temp", case=False, na=False)
    day_features = shap_imp.loc[day_mask]
    day_sum = day_features["Mean_|SHAP|"].sum()
    
    nce_rows = shap_imp.loc[shap_imp["Feature"].str.contains("Nighttime Cooling Efficiency", na=False)]
    nce_val = nce_rows["Mean_|SHAP|"].values[0] if not nce_rows.empty else np.nan
    
    night_day_ratio = night_sum / day_sum if day_sum > 0 else np.nan
    
    print(f"Nighttime total |SHAP| (sum of night features): {night_sum:.6f}")
    print(f"Daytime total |SHAP| (sum of day features): {day_sum:.6f}")
    print(f"Night/Day ratio: {night_day_ratio:.2f}")
    print(f"NCE (top feature): {nce_val:.6f}")

# ====================================================================
# 5. FOREST BUFFERING (From later code, with fixes)
# ====================================================================
print(f"\n\n{'='*80}")
print("SECTION 3: FOREST BUFFERING CLAIM VERIFICATION")
print("="*80)

if X_sample is not None and shap_vals is not None:
    LST_FEAT = "Daytime Surface Temp (Mean)"
    FOREST_FEAT = "Deciduous Forest %"
    
    if LST_FEAT in fn_list and FOREST_FEAT in fn_list:
        lst_idx = fn_list.index(LST_FEAT)
        lst_x = X_sample[LST_FEAT].values
        lst_shap = shap_vals[:, lst_idx]
        forest_x = X_sample[FOREST_FEAT].values
        
        tq = pd.qcut(lst_x, q=4, labels=["Q1", "Q2", "Q3", "Q4"])
        fcat = pd.cut(forest_x, bins=[-np.inf, 0.05, 0.20, np.inf], labels=["Low (<5%)", "Medium (5-20%)", "High (>20%)"])
        
        df_i = pd.DataFrame({"tq": tq, "fc": fcat, "sv": lst_shap}).dropna()
        
        q4l = df_i[(df_i["fc"] == "Low (<5%)") & (df_i["tq"] == "Q4")]["sv"].median()
        q4h = df_i[(df_i["fc"] == "High (>20%)") & (df_i["tq"] == "Q4")]["sv"].median()
        gap = q4l - q4h
        buffering_reduction = abs(gap / q4l) * 100 if q4l != 0 else 0
        
        print(f"Q4 Low Forest SHAP penalty: +{q4l:.3f}")
        print(f"Q4 High Forest SHAP penalty: +{q4h:.3f}")
        print(f"Attenuation gap: {gap:.3f}")
        print(f"Forest buffering reduction: {buffering_reduction:.1f}%")

# ====================================================================
# 6. SOIL OPTIMAL BAND (From later code, with fixes)
# ====================================================================
print(f"\n\n{'='*80}")
print("SECTION 4: SOIL MOISTURE OPTIMAL BAND")
print("="*80)

if df is not None:
    soil_col = "Soil_mean"
    if soil_col in df.columns and "MeanLifeExpectency" in df.columns:
        df_s = df[[soil_col, "MeanLifeExpectency"]].dropna()
        df_s = df_s[(df_s[soil_col] > df_s[soil_col].quantile(0.01)) & (df_s[soil_col] < df_s[soil_col].quantile(0.99))]
        
        order = np.argsort(df_s[soil_col].values)
        lw = lowess(df_s["MeanLifeExpectency"].values[order], df_s[soil_col].values[order], frac=0.20, return_sorted=True)
        lx, ly = lw[:, 0], lw[:, 1]
        
        soil_peak_idx = np.argmax(ly)
        soil_peak = float(lx[soil_peak_idx])
        
        print(f"Optimal soil moisture (peak LE from LOWESS on CSV data): {soil_peak:.2f}")

# ====================================================================
# 7. LIVESTOCK BREAKPOINTS (Fixed)
# ====================================================================
print(f"\n\n{'='*80}")
print("SECTION 5: AGRICULTURAL / LIVESTOCK BREAKPOINTS")
print("="*80)

if df is not None:
    cattle_col = "mean_cattle"
    if cattle_col in df.columns and "MeanLifeExpectency" in df.columns:
        df_c = df[[cattle_col, "MeanLifeExpectency"]].dropna()
        df_c_grouped = df_c.groupby(cattle_col, as_index=False)["MeanLifeExpectency"].mean().sort_values(cattle_col)
        
        x = df_c_grouped[cattle_col].values
        y = df_c_grouped["MeanLifeExpectency"].values
        
        try:
            f = interp1d(x, y, kind="cubic", fill_value="extrapolate")
        except TypeError:
            f = interp1d(x, y, kind="linear", fill_value=np.nan)
        x_fine = np.linspace(x.min(), x.max(), 500)
        y_fine = f(x_fine)
        
        dy = np.gradient(y_fine, x_fine)
        d2y = np.gradient(dy, x_fine)
        inflection_idx = np.argmin(np.abs(d2y))
        inflection_cattle = x_fine[inflection_idx]
        
        print(f"Cattle density breakpoint (inflection from cubic interp on CSV data): {inflection_cattle:.0f}")

# ====================================================================
# 8. TEMPORAL STABILITY (From later code)
# ====================================================================
print(f"\n\n{'='*80}")
print("SECTION 6: TEMPORAL STABILITY")
print("="*80)

t3 = tables.get("table3_temporal")
if t3 is not None:
    years = t3.index.values
    r2_vals = t3["R2"].values
    mae_vals = t3["MAE"].values
    rmse_vals = t3["RMSE"].values
    
    sl_r2, _, _, _, _ = linregress(years, r2_vals)
    sl_mae, _, _, _, _ = linregress(years, mae_vals)
    sl_rmse, _, _, _, _ = linregress(years, rmse_vals)
    
    print(f"R² trend slope: {sl_r2:+.6f} per year")
    print(f"MAE trend slope: {sl_mae:+.6f} per year")
    print(f"RMSE trend slope: {sl_rmse:+.6f} per year")

# ====================================================================
# 9. BIAS & RESIDUALS (From later code, with fixes)
# ====================================================================
print(f"\n\n{'='*80}")
print("SECTION 7: BIAS & RESIDUALS ANALYSIS")
print("="*80)

if predictions is not None:
    predictions["fips"] = predictions["fips"].astype(str).str.zfill(5)
    res_by_fips = predictions.groupby("fips")["residual"].mean()
    
    over = res_by_fips.nsmallest(5)
    under = res_by_fips.nlargest(2)
    
    print("Worst over-predicted (most negative residuals):")
    for fips, val in over.items():
        print(f"  FIPS {fips}: {val:+.4f}")
    
    print("Worst under-predicted (most positive residuals):")
    for fips, val in under.items():
        print(f"  FIPS {fips}: {val:+.4f}")
    
    res_arr = predictions["residual"].values
    res_mean = res_arr.mean()
    res_std = res_arr.std()
    res_skew = skew(res_arr)
    res_min, res_max = res_arr.min(), res_arr.max()
    res_q25, res_q50, res_q75 = np.percentile(res_arr, [25, 50, 75])
    
    print(f"Residual stats (from predictions): Mean={res_mean:+.6f}, Std={res_std:.6f}, Skew={res_skew:.4f}, Min={res_min:.4f}, Q50={res_q50:.4f}, Max={res_max:.4f}")

# ====================================================================
# 10. MORAN'S I (Fixed for contiguous US and attribute issues)
# ====================================================================
print(f"\n\n{'='*80}")
print("SECTION 8: SPATIAL AUTOCORRELATION (Moran's I)")
print("="*80)

if HAS_SPATIAL and predictions is not None and COUNTY_SHP.exists():
    try:
        gdf = gpd.read_file(COUNTY_SHP)
        gdf = gdf[gdf['STATEFP'].isin(contiguous_states)]  # Contiguous US only
        gdf["fips"] = gdf["GEOID"].astype(str).str.zfill(5)
        
        gdf = gdf.merge(res_by_fips.rename("res_mean"), left_on="fips", right_index=True, how="inner")
        gdf = gdf.to_crs("EPSG:5070")
        
        w = libpysal.weights.Queen.from_dataframe(gdf)
        if hasattr(w, 'transform'):
            w.transform = "r"  # Fixed: check hasattr to avoid assignment error
        
        mi = esda.moran.Moran(gdf["res_mean"].values, w)
        
        # Safely extract attributes
        moran_i = mi.I
        ei = mi.EI
        p_sim = getattr(mi, 'p_sim', 1.0)
        z_score = getattr(mi, 'z_norm', 0.0)
        
        print(f"Moran's I: {moran_i:.6f} (expected: {ei:.6f})")
        print(f"Z-score: {z_score:.4f}, P-value (sim): {p_sim:.6f}")
        if p_sim < 0.05:
            print("Significant spatial autocorrelation detected.")
        else:
            print("No significant spatial autocorrelation.")
    except Exception as e:
        print(f"Moran's I failed: {e}")

# ====================================================================
# 11. LOWESS INFLECTION POINTS (Top 10 Features, Fixed)
# ====================================================================
print(f"\n\n{'='*80}")
print("SECTION 9: LOWESS CURVE ANALYSIS - INFLECTION POINTS")
print("="*80)

def get_lowess_inflection(feat_name):
    if feat_name not in fn_list or shap_vals is None or X_sample is None:
        return np.nan
    idx = fn_list.index(feat_name)
    x = X_sample[feat_name].values
    y = shap_vals[:, idx]
    valid = np.isfinite(x) & np.isfinite(y)
    x, y = x[valid], y[valid]
    if len(x) < 10:
        return np.nan
    order = np.argsort(x)
    lw = lowess(y[order], x[order], frac=0.20, return_sorted=True)
    lx, ly = lw[:, 0], lw[:, 1]
    dy = np.gradient(ly, lx)
    sign_changes = np.where(np.diff(np.sign(dy)))[0]
    return float(lx[sign_changes[0]]) if len(sign_changes) > 0 else np.nan

if shap_imp is not None:
    top10_features = shap_imp.nlargest(10, "Mean_|SHAP|")
    for _, row in top10_features.iterrows():
        feat = row["Feature"]
        inflection = get_lowess_inflection(feat)
        print(f"{feat}: Inflection at {inflection:.4f}" if not np.isnan(inflection) else f"{feat}: No inflection found")

# ====================================================================
# 12. PHYSICAL THRESHOLDS (Combined, Fixed)
# ====================================================================
print(f"\n\n{'='*80}")
print("SECTION 10: PHYSICAL THRESHOLDS")
print("="*80)

# Temperature thresholds
night_feat = "Nighttime Temp (10th Percentile)"
day_feat = "Daytime Temp (10th Percentile)"
night_inflection = get_lowess_inflection(night_feat)
day_inflection = get_lowess_inflection(day_feat)
print(f"Night temp inflection: {night_inflection:.2f}°C" if not np.isnan(night_inflection) else "Night temp: N/A")
print(f"Day temp inflection: {day_inflection:.2f}°C" if not np.isnan(day_inflection) else "Day temp: N/A")

# ====================================================================
# 13. SAVE VERIFICATION REPORT
# ====================================================================
print(f"\n\n{'='*80}")
print("SECTION 11: SAVING FINAL REPORT")
print("="*80)

report = {
    "overall_r2": r2 if 'r2' in locals() else np.nan,
    "overall_mae": mae if 'mae' in locals() else np.nan,
    "overall_rmse": rmse if 'rmse' in locals() else np.nan,
    "night_day_ratio": night_day_ratio if 'night_day_ratio' in locals() else np.nan,
    "nce_shap": nce_val if 'nce_val' in locals() else np.nan,
    "forest_buffering_pct": buffering_reduction if 'buffering_reduction' in locals() else np.nan,
    "soil_optimal_band": soil_peak if 'soil_peak' in locals() else np.nan,
    "cattle_inflection": inflection_cattle if 'inflection_cattle' in locals() else np.nan,
}

pd.Series(report).round(4).to_csv(RESULTS_DIR / "paper_verification_report.csv")
print("✅ Full verification report saved.")
```
